```python
# 1. Pandas version

import pandas as pd

pd.__version__
```

```python
# 2. Records count

df = pd.read_csv('car_fuel_efficiency_2026.csv')
df.shape[0]
```

```python
# 3. Fuel types

df.fuel_type.values.unique().size
```

```python
# 4. Missing values

df.isnull().any().sum()
```

```python
# 5. Max fuel efficiency

df[df['origin'] == 'Asia'].fuel_efficiency_mpg.max()
```

```python
# 6. Median value of horsepower

median_before = df["horsepower"].median()
print(median_before)

most_frequent = df["horsepower"].mode()[0]
print(most_frequent)

df["horsepower_filled"] = df["horsepower"].fillna(most_frequent)

median_after = df["horsepower_filled"].median()
print(median_after)
```

```python
# 7. Sum of weights

import numpy as np

values = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
X = values.to_numpy()
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
w.sum()
```
